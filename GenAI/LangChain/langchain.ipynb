# LangChain Framework - Beginner's Guide

## What is LangChain?

LangChain is a framework that simplifies building applications with Large Language Models (LLMs). It provides:
- **Standardized interfaces** for working with different LLMs (GPT, Claude, Gemini, etc.)
- **Memory management** to maintain conversation context
- **Chains** to build complex workflows
- **Agents** for autonomous decision-making
- **Tools** for external integrations (search, databases, APIs)

## Why Use LangChain?

✅ Makes LLM integration simple and modular
✅ Reduces boilerplate code
✅ Enables composable AI workflows
✅ Handles prompt management efficiently
✅ Supports multiple LLM providers

## Detailed Uses of LangChain Modules

LangChain is organized into modular components so you can build LLM applications in a clean and reusable way. Below is a practical guide to the main modules and what each one is used for.

### 1. `langchain_core`
This is the foundation of LangChain. It contains the base abstractions used across the framework.

#### Common uses:
- Define prompts, messages, chains, and output formats
- Work with provider-independent building blocks
- Build reusable components that can connect to any supported LLM

#### Key parts:
- **Prompt templates**: Create reusable prompts with variables
- **Messages**: Represent chat turns like human and AI messages
- **Runnables**: Compose workflows using the pipe `|` operator
- **Output parsers**: Convert raw model output into structured data

## Core Components of LangChain

### 1. **Prompts (PromptTemplate)**
A prompt is the instruction/question you send to an LLM. LangChain provides templates to make prompts reusable and dynamic.

**Use case**: Instead of hardcoding prompts, use templates with variables that can be filled dynamically.

````python
# Example 1: Creating a Simple PromptTemplate
from langchain_core.prompts import PromptTemplate

# Create a reusable prompt template
prompt = PromptTemplate(
    input_variables=["topic", "level"],
    template="Explain {topic} to a {level} student in 2-3 sentences."
)

# Use the template with different inputs
print("=" * 60)
print("PROMPT 1 (Beginner):")
print(prompt.format(topic="Machine Learning", level="beginner"))
print()
print("PROMPT 2 (Advanced):")
print(prompt.format(topic="Machine Learning", level="PhD"))
````

### 2. **Language Models (LLMs)**
An LLM is the core AI component. LangChain provides a unified interface for different LLM providers.

**Use case**: Write once, switch between GPT, Claude, Gemini without changing your code structure.

````python
# Example 2.1: Using Gemini API with LangChain
print("=" * 70)
print("OPTION 1: GEMINI API (Cloud LLM)")
print("=" * 70)

# Installation: pip install langchain-google-genai

from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

load_dotenv()

# Initialize Gemini model
gemini_llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    temperature=0.7,
    api_key=os.getenv("GEMINI_API_KEY")
)

# Simple invocation
print("\n🤖 Calling Gemini (gemini-2.5-flash-lite)...")
response = gemini_llm.invoke("What is Machine Learning in simple terms?")
print(f"\n📝 Response:\n{response.content[:200]}...")  # Show first 200 chars

print("\n" + "=" * 70)
print("OPTION 2: OLLAMA (Local LLM - Free & Private)")
print("=" * 70)

# Installation: 
# 1. Download Ollama from https://ollama.ai
# 2. Run: ollama pull llama2 (or mistral, neural-chat, etc.)
# 3. Install: pip install langchain-ollama

from langchain_ollama import ChatOllama

try:
    # Initialize Ollama model (runs locally)
    ollama_llm = ChatOllama(
        model="llama3.2",  # Can also use: mistral, neural-chat, dolphin-mixtral
        base_url="http://localhost:11434",  # Default Ollama server
        temperature=0.7
    )
    
    print("\n✅ Ollama server is running!")
    print("🤖 Calling Ollama (llama3.2 - runs locally)...")
    response = ollama_llm.invoke("What is Machine Learning in simple terms?")
    print(f"\n📝 Response:\n{response.content[:200]}...")
    
except Exception as e:
    print(f"\n⚠️ Ollama not available: {e}")
    print("   Make sure Ollama is installed and running on http://localhost:11434")
    print("   Download from: https://ollama.ai")

print("\n" + "=" * 70)
print("OPTION 3: OTHER CLOUD LLMs")
print("=" * 70)

print("""
# OpenAI (GPT-4, GPT-3.5)
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4", temperature=0.7, api_key="sk-...")

# Anthropic (Claude)
from langchain_anthropic import ChatAnthropic
llm = ChatAnthropic(model="claude-3-sonnet-20240229", temperature=0.7)

# HuggingFace (Open-source models)
from langchain_huggingface import HuggingFaceEndpoint
llm = HuggingFaceEndpoint(repo_id="mistralai/Mistral-7B-Instruct-v0.1")

# More options at: https://python.langchain.com/docs/integrations/llms/
""")
````

## Quick Comparison: Choosing the Right LLM

| LLM | Type | Cost | Speed | Privacy | Setup |
|-----|------|------|-------|---------|-------|
| **Gemini 2.5 Flash** | Cloud | 💰 Pay-as-you-go | ⚡ Very Fast | ❌ Cloud | Easy |
| **Ollama (Llama3.2)** | Local | ✅ Free | 🐢 Slower* | ✅ Private | Medium |
| **GPT-4** | Cloud | 💰💰💰 Expensive | ⚡ Fast | ❌ Cloud | Easy |
| **Claude 3** | Cloud | 💰💰 Moderate | ⚡ Fast | ❌ Cloud | Easy |
| **HuggingFace** | Cloud/Local | 💰 Free-Paid | ⚡⚡ Fast | Varies | Medium |

*Speed depends on your hardware

### When to Use Each:
- **Gemini**: Best balance of speed, cost, and quality
- **Ollama**: When you want free, private, no API keys
- **GPT-4**: Premium quality for complex tasks
- **Claude**: Great for analysis and reasoning
- **HuggingFace**: Open-source with many model choices

````python
# Example 2.2: Universal LLM Wrapper Function
# This shows how to switch between different LLMs easily

from typing import Literal
import os
from dotenv import load_dotenv

load_dotenv()

def get_llm(model_type: Literal["gemini", "ollama", "openai"] = "gemini"):
    """
    Get different LLM instances based on model_type.
    This makes it easy to switch models without changing your main code!
    """
    
    if model_type == "gemini":
        from langchain_google_genai import ChatGoogleGenerativeAI
        return ChatGoogleGenerativeAI(
            model="gemini-2.5-flash-lite",
            temperature=0.7,
            api_key=os.getenv("GEMINI_API_KEY")
        )
    
    elif model_type == "ollama":
        from langchain_ollama import ChatOllama
        return ChatOllama(
            model="llama2",
            base_url="http://localhost:11434",
            temperature=0.7
        )
    
    elif model_type == "openai":
        from langchain_openai import ChatOpenAI
        return ChatOpenAI(
            model="gpt-4",
            temperature=0.7,
            api_key=os.getenv("OPENAI_API_KEY")
        )
    
    else:
        raise ValueError(f"Unknown model type: {model_type}")


print("=" * 70)
print("UNIVERSAL LLM WRAPPER - Easy Model Switching")
print("=" * 70)

# Example usage - Try different models
for model in ["gemini", "ollama", "openai"]:
    try:
        print(f"\n🔄 Testing {model.upper()}...")
        llm = get_llm(model)
        print(f"✅ {model.upper()} LLM initialized successfully")
        print(f"   Type: {type(llm).__name__}")
    except Exception as e:
        print(f"⚠️ {model.upper()} not available: {str(e)[:50]}...")

print("\n" + "=" * 70)
print("💡 Tip: This wrapper pattern makes your code reusable!")
print("   Change just ONE LINE to switch between ALL these models.")
print("=" * 70)
````

````python
# Example 2.3: Complete Workflow - PromptTemplate + Gemini LLM

from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

load_dotenv()

print("=" * 70)
print("COMPLETE WORKFLOW: Prompt Template → Gemini LLM")
print("=" * 70)

# Step 1: Create a prompt template
explain_prompt = PromptTemplate(
    input_variables=["topic", "audience"],
    template="Explain {topic} to a {audience} in 2-3 sentences. Keep it simple and engaging."
)

# Step 2: Initialize Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    temperature=0.7,
    api_key=os.getenv("GEMINI_API_KEY")
)

# Step 3: Create a chain (Prompt → LLM)
chain = explain_prompt | llm

# Step 4: Use the chain
print("\n🔗 Creating chain: PromptTemplate → Gemini LLM")

try:
    # Example 1
    print("\n👾 Example 1: AI for Beginners")
    print("-" * 70)
    formatted_prompt = explain_prompt.format(topic="Artificial Intelligence", audience="5th grader")
    print(f"Generated prompt:\n{formatted_prompt}\n")
    
    response = chain.invoke({"topic": "Artificial Intelligence", "audience": "5th grader"})
    print(f"Response:\n{response.content}\n")
    
    # Example 2
    print("\n🤖 Example 2: ML for PhD Student")
    print("-" * 70)
    formatted_prompt = explain_prompt.format(topic="Machine Learning", audience="PhD student in Computer Science")
    print(f"Generated prompt:\n{formatted_prompt}\n")
    
    response = chain.invoke({"topic": "Machine Learning", "audience": "PhD student in Computer Science"})
    print(f"Response:\n{response.content}\n")
    
except Exception as e:
    print(f"⚠️ Error: {e}")
    print("Make sure GEMINI_API_KEY is set in your .env file")

print("=" * 70)
print("✅ This is how LangChain chains work!")
print("   Prompt Template formats input → LLM processes it → Returns result")
print("=" * 70)
````

### 3. **Chains**
Chains combine prompts and LLMs to build workflows. The simplest chain is `prompt -> LLM`.

**Use case**: Build sequential workflows where output of one step becomes input to the next.

**Chain Benefits:**
- Modular and reusable components
- Easy to debug each step
- Composable for complex applications

````python
# Example 3: Building a Simple Chain
from langchain_core.prompts import PromptTemplate

# Step 1: Create a prompt template
topic_prompt = PromptTemplate(
    input_variables=["topic"],
    template="Write a short quiz question about {topic}:"
)

print("=" * 60)
print("SIMPLE CHAIN DEMONSTRATION")
print("=" * 60)

# Step 2: This would be combined with LLM in real usage
# chain = topic_prompt | llm
# response = chain.invoke({"topic": "Python Programming"})

# For now, let's show what the prompt produces:
print("\nPrompt Output:")
print(topic_prompt.format(topic="Python Programming"))
print("\nIn a real chain, this prompt would be sent to an LLM like GPT-4")
print("and the LLM would generate a response.")
````

### 4. **Memory**
Memory allows your application to remember previous conversations and context.

**Types of Memory:**
- **Buffer Memory**: Stores last N messages
- **Summary Memory**: Summarizes old messages to save tokens
- **Entity Memory**: Remembers specific facts about entities
- **Conversation Memory**: Specialized for multi-turn dialog

**Use case**: Build chatbots that remember conversation history and provide contextual responses.

````python
# Example 4: Simple Memory Implementation
from langchain_core.messages import HumanMessage, AIMessage

# Simulating a conversation with memory
conversation_history = []

def chat(user_message):
    """Simulate a chat with memory"""
    # Add user message to history
    conversation_history.append(HumanMessage(content=user_message))
    print(f"User: {user_message}")
    
    # In a real app, the LLM would see all history
    print(f"  [Memory has {len(conversation_history)} messages]")
    print()

print("=" * 60)
print("MEMORY DEMONSTRATION")
print("=" * 60)
print()

# Simulate a multi-turn conversation
chat("Hi, my name is Alice")
chat("What did I just tell you?")
chat("Can you remember my name?")

print("Memory allows the model to maintain context across multiple turns")
````

### 5. **Agents**
Agents are autonomous systems that can decide which tools to use and in what order to achieve a goal.

**Key Components:**
- **LLM**: Makes decisions
- **Tools**: Actions the agent can take (calculators, search, APIs, etc.)
- **Memory**: Remembers what it's done
- **Executor**: Runs the agent loop

**Use case**: Build an AI assistant that can search the web, calculate, query databases autonomously to answer questions.

````python
# Example 5: Understanding Agent Tools
print("=" * 60)
print("AGENTS & TOOLS CONCEPT")
print("=" * 60)
print()

# Define some tools an agent might use
tools = {
    "calculator": "Can perform mathematical calculations",
    "search": "Can search the internet for information",
    "database_query": "Can query a database",
    "email_sender": "Can send emails",
    "api_caller": "Can call external APIs"
}

print("Available Tools:")
for tool_name, description in tools.items():
    print(f"  • {tool_name}: {description}")

print()
print("Agent Workflow Example:")
print("  User Question: 'What's the population of France and multiply by 2?'")
print("  1. Agent reads question")
print("  2. Decides tools needed: 'search' + 'calculator'")
print("  3. Uses 'search' tool → finds population = 67 million")
print("  4. Uses 'calculator' tool → 67M × 2 = 134M")
print("  5. Returns answer to user")
print()
print("This is what makes agents powerful - they reason about which")
print("tools to use based on the task!")
````

## Best Practices for LangChain Development

### ✅ DO's:
1. **Use environment variables** for API keys (`.env` file)
2. **Create reusable prompt templates** instead of hardcoding prompts
3. **Add error handling** for API calls and model errors
4. **Monitor token usage** to manage costs
5. **Test chains separately** before combining them
6. **Use appropriate memory types** based on use case
7. **Log conversations** for debugging and improvement

### ❌ DON'Ts:
1. **Don't hardcode API keys** in your code
2. **Don't create long chains** without testing individual steps
3. **Don't ignore rate limits** of LLM providers
4. **Don't store unlimited history** in memory (it's expensive)
5. **Don't assume one LLM works** for all scenarios

## Super Simple LangChain Modules (Quick Practice)

Run these cells one by one. Each cell teaches one module with the smallest possible example.

````python
# 1) PromptTemplate module
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("Explain {topic} in one line.")
print(prompt.format(topic="Neural Networks"))
````

````python
# 2) Chat Model module (Gemini)
# pip install langchain-google-genai python-dotenv

import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

gemini = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    temperature=0.3,
    api_key=os.getenv("GEMINI_API_KEY"),
)

reply = gemini.invoke("Say hello in 1 sentence.")
print(reply.content)
````

````python
# 3) Chat Model module (Ollama local)
# pip install langchain-ollama
# Make sure Ollama is running and a model exists: ollama pull llama3.2

from langchain_ollama import ChatOllama

try:
    local_llm = ChatOllama(model="llama3.2", base_url="http://localhost:11434")
    out = local_llm.invoke("What is AI in one line?")
    print(out.content)
except Exception as e:
    print("Ollama not ready:", e)
````

````python
# 4) Chain module (Prompt -> LLM)
from langchain_core.prompts import PromptTemplate

chain_prompt = PromptTemplate.from_template("Explain {topic} for a beginner.")

# Use Gemini chain; change to local_llm if you want local model
chain = chain_prompt | gemini

result = chain.invoke({"topic": "LangChain"})
print(result.content)
````

````python
# 5) Output Parser module
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
parsed_chain = chain_prompt | gemini | parser

text = parsed_chain.invoke({"topic": "Embeddings"})
print(type(text))
print(text)
````

````python
# 6) Memory module (simple conversation history)
from langchain_core.messages import HumanMessage, AIMessage

chat_history = []
chat_history.append(HumanMessage(content="My name is Hamen"))
chat_history.append(AIMessage(content="Nice to meet you, Hamen"))
chat_history.append(HumanMessage(content="What is my name?"))

for m in chat_history:
    print(f"{m.type}: {m.content}")
````

````python
# 7) Tools module (convert a Python function into a tool)
from langchain_core.tools import tool

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two integers."""
    return a * b

print("Tool name:", multiply.name)
print("Tool call result:", multiply.invoke({"a": 6, "b": 7}))
````

````python
# 8) Agent-style reasoning (simple pattern)
# Full production agents need extra setup, but this shows the core idea.

question = "What is 25 * 4?"

if "*" in question or "multiply" in question.lower():
    answer = multiply.invoke({"a": 25, "b": 4})
    print("Used tool: multiply")
    print("Answer:", answer)
else:
    reply = gemini.invoke(question)
    print("Used model directly")
    print(reply.content)
````
